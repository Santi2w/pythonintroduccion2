```python
# Creación aleatoria de datos
import numpy as np
import random
random.seed(42)
```

```python
# Creo una variable de tipo lista vacía
nivel_estudios = []
# Creo o ingreso a mi lista la variable Secuandaria y creo 110 veces secundaria,
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)
# Mezcla y repite los datos de forma randomica
nivel_estudios = np.random.permutation(nivel_estudios)
```

```python
print(nivel_estudios)
```

```python
# DIAGRAMA DE BARRAS
import matplotlib.pyplot as plt
import numpy as np
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

```python
plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras')
plt.bar(estudios, frecuencia) # https://matplotlib.org/stable/api/_as_gen/matplo
plt.show()
```

```python
# DIAGRAMA DE SECTORES
colores = {'red', 'blue', 'orange', 'green', 'cyan'}
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```

```python
# Tarea implementar con pandas esta información
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Creación de los datos iniciales (reproduciendo las frecuencias del documento)
nivel_estudios = []
nivel_estudios.extend(['Secundaria'] * 110)
nivel_estudios.extend(['Bachillerato'] * 172)
nivel_estudios.extend(['Grado'] * 145)
nivel_estudios.extend(['Master'] * 52)
nivel_estudios.extend(['Doctorado'] * 21)

# Permutación aleatoria para mantener la consistencia con el ejercicio
np.random.seed(42)  # Fijamos la semilla de numpy para reproducibilidad
nivel_estudios_perm = np.random.permutation(nivel_estudios)

# 2. Crear el DataFrame de Pandas
df = pd.DataFrame(nivel_estudios_perm, columns=['Nivel de estudios'])

# 3. Calcular las frecuencias absolutas de manera simple con value_counts()
# Usamos .sort_index() para que se ordenen alfabéticamente igual que en tu PDF
frecuencias = df['Nivel de estudios'].value_counts().sort_index()

# Mostramos la tabla de frecuencias en el notebook
print("TABLA DE FRECUENCIAS")
print(frecuencias)
print("\n")
```

```python
# 4. DIAGRAMA DE BARRAS CON PANDAS
plt.figure(figsize=(8, 5))
frecuencias.plot(kind='bar', color='tab:blue', edgecolor='black')
plt.title('Diagrama de barras')
plt.xlabel('Nivel de estudios')
plt.ylabel('Frecuencia absoluta')
plt.xticks(rotation=0)  # Mantiene las etiquetas horizontales
plt.show()
```
